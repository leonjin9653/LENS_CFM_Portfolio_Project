```python
import yfinance as yf
import pandas as pd
```

```python
# Pulls the return on equity for a stock
def pull_roe(ticker_symbol):
    ticker = yf.Ticker(ticker_symbol)

    info = ticker.info
    roe = info.get("returnOnEquity")

    if roe != None:
        roe = roe * 100

    return roe


# Test with a few stocks
test_tickers = ["MSFT", "JNJ", "NVDA", "JPM"]

roe_data = []

for symbol in test_tickers:
    roe = pull_roe(symbol)

    roe_data.append({
        "Ticker": symbol,
        "ROE (%)": roe
    })

roe_data = pd.DataFrame(roe_data)

display(roe_data)
```
