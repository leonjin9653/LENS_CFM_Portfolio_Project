```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python
ticker = "NVDA"

start_date = "2025-01-01" #variable start and end date. Probably shouldn't be hard coded but can be adjusted later.
end_date = "2025-03-01" 

data = yf.download(ticker, start=start_date, end=end_date)
```

```python
def price_std(data):
    std = (data["Close"]).std()
    return (std)

std = price_std(data)
print(std)
    
```

```python
def avg_price_std(data):

    biweekly_std = data["Close"].resample("2W").std()

    avg_std = biweekly_std.dropna().mean()

    return avg_std


output = avg_price_std(data)

print(output)
```
