```python
import yfinance as yf
import pandas as pd
```

```python
# Pulls trailing and forward EPS for a stock
def pull_eps(ticker_symbol):
    ticker = yf.Ticker(ticker_symbol)

    info = ticker.info

    trailing_eps = info.get("trailingEps")
    forward_eps = info.get("forwardEps")

    return trailing_eps, forward_eps


# Test with a few stocks
test_tickers = ["MSFT", "JNJ", "NVDA", "JPM"]

eps_data = []

for symbol in test_tickers:
    trailing_eps, forward_eps = pull_eps(symbol)

    eps_data.append({
        "Ticker": symbol,
        "Trailing EPS": trailing_eps,
        "Forward EPS": forward_eps
    })

eps_data = pd.DataFrame(eps_data)

display(eps_data)
```
