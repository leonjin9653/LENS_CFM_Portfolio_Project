The ATR will output us two different values. The ATR helps us measure how much its price moves. This way we can track ATR with higher price movements by percentage and also by dollar value. This can also tell us if the stock trades at a high price. The ATR returns an ATR percentage as well as an ATR dollar value.
So when we are deciding how to use the ATR, we can have like a baseline % based on which portfolio.

```python
from IPython.display import display, Math, Latex

import pandas as pd
import numpy as np
import numpy_financial as npf
import yfinance as yf
import matplotlib.pyplot as plt
import random
from datetime import datetime
```

```python

# Calculates ATR and ATR% for a stock
def calculate_atr(stock_data):
    high = stock_data["High"]
    low = stock_data["Low"]
    prev_close = stock_data["Close"].shift(1)

    true_range = pd.concat([
        high - low,
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)

    atr = true_range.tail(14).mean()
    atr_percent = atr / stock_data["Close"].iloc[-1] * 100

    return atr, atr_percent
```

```python

# For testing purposes
test_tickers = ["MSFT", "JNJ", "PG", "NVDA", "JPM", "NEE"]

results = []

for ticker in test_tickers:
    data = yf.Ticker(ticker).history(
        start="2026-05-01",
        end="2026-11-18",
        auto_adjust=False
    )

    atr, atr_percent = calculate_atr(data)

    results.append({
        "Ticker": ticker,
        "ATR": atr,
        "ATR_Percent": atr_percent
    })

atr_data = pd.DataFrame(results)

display(atr_data.sort_values("ATR_Percent"))
```
