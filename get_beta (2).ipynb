```python
import yfinance as yf
import pandas as pd

```

```python
# Pulls the beta for a stock
def pull_beta(ticker_symbol):
    ticker = yf.Ticker(ticker_symbol)

    info = ticker.info
    beta = info.get("beta")

    return beta


# Test with a few stocks
test_tickers = ["MSFT", "JNJ", "NVDA", "JPM"]

beta_data = []

for symbol in test_tickers:
    beta = pull_beta(symbol)

    beta_data.append({
        "Ticker": symbol,
        "Beta": beta
    })

beta_data = pd.DataFrame(beta_data)

display(beta_data)
```
