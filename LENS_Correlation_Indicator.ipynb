### Correlation Indicator

This function calculates the correlation between stocks based on their past daily returns in CAD. The idea is to see which stocks tend to move together so we can avoid having too many stocks that behave similarly.

For the low-risk portfolio, we'll first pick a stock using our other indicators, then use correlation to help decide which stocks to add next. We'll calculate the average correlation between each potential stock and the stocks we've already selected, and generally favour lower correlations to help diversify the portfolio and reduce risk.

Obviously, correlation isn't everything, so we'll still consider our other indicators before choosing a stock. For the growth portfolio, we can use correlation more as a secondary factor when two stocks have similar rankings.

Once we combine everyone's code, we'll implement this into our main stock-selection process.

```python
from IPython.display import display, Math, Latex

import pandas as pd
import numpy as np
import numpy_financial as npf
import yfinance as yf
import matplotlib.pyplot as plt
import random
from datetime import datetime
```

```python
#This code is considering that we do all the preliminary downloads and test cases of eligible stocks before running the function
# Calculates the correlation between stocks
def calculate_correlations(prices):
    returns = prices.pct_change(fill_method=None)
    return returns.corr()


# Calculates the average correlation with selected stocks
def average_correlation(candidate, selected_stocks, correlations):
    return correlations.loc[candidate, selected_stocks].mean()

```

```python
#I did all the downloads and the exchange rates that would happen in the main just for testing purposes
# For testing purposes
test_tickers = [
    "AAPL", "MSFT", "NVDA", "GOOGL", "AMZN",
    "JPM", "BAC", "V", "JNJ", "PG",
    "KO", "PEP", "WMT", "XOM", "CVX"
]

# Download historical prices
prices = yf.download(
    test_tickers,
    start="2026-05-01",
    end="2026-11-18",
    auto_adjust=False
)["Close"]

# Convert USD prices to CAD
exchange_rate = yf.download(
    "CAD=X",
    start="2026-05-01",
    end="2026-11-18"
)["Close"].squeeze()

# Matches the exchange rates to the stock dates, fills in missing rates, and converts the prices from USD to CAD
prices = prices.mul(exchange_rate.reindex(prices.index).ffill(), axis=0)

# Calculate correlations
correlations = calculate_correlations(prices)

display(correlations)

# Test average correlation
print(average_correlation("JPM", ["MSFT", "JNJ", "PG"], correlations))
```
